<h1> Rataan Bergerak (Moving Average) </h1>

Diberikan suatu deret waktu ${Y_t}$, dan deret white noise yang tidak teramati ${e_t}$, yaitu barisan peubah acak yang saling bebas dan memiliki sebaran identik dengan nilai tengah nol.

Proses rataan bergerak (moving average) secara umum atau MA (q) dapat dinyatakan dalam model sebagai berikut:

$$
Y_t = e_t - \theta_1 e_{t-1} - \theta_2 e_{t-2} - \cdots - \theta_q e_{t-q}
$$

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.arima_process import ArmaProcess
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf

import warnings
warnings.filterwarnings("ignore")
```

## MA (1)

Proses MA(1) memiliki persamaan:

$$
Y_t = e_t - \theta e_{t-1}
$$
Kita akan membuat simulasi proses MA(1) dengan $\theta=-0.9$.

```python
np.random.seed(1302)
ar1 = np.array([1])
ma1 = np.array([1, 0.9]) #nilai parameter theta dalam sintaks dituliskan sebagai negatif dari nilai parameter asli, misal parameter asli -0.9 dalam sintaks ditulis 0.9
ma_object1 = ArmaProcess(ar1,ma1)
ma1 = ma_object1.generate_sample(nsample=120)
```

```python
plt.figure()
plt.plot(ma1)
plt.xlabel('Waktu'), plt.ylabel(r'$Y_t$')
plt.show()
```

Bagaimana interpretasi Anda mengenai plot di atas?

Untuk melihat apakah terdapat hubungan dalam deret antar waktu kita lihat dari plot pencar data berpasangan.

```python
plt.figure(figsize=(6,5))
pd.plotting.lag_plot(pd.DataFrame(ma1), lag=1, c='k')
plt.xlabel(r'$Y_{t-1}$'),plt.ylabel(r'$Y_{t}$')
plt.show()
```

Apa interpretasi Anda?

```python
plt.figure(figsize=(6,5))
pd.plotting.lag_plot(pd.DataFrame(ma1), lag=2, c='k')
plt.xlabel(r'$Y_{t-2}$'),plt.ylabel(r'$Y_{t}$')
plt.show()
```

Bagaimana interpretasi Anda dari plot-plot di atas?

Selain menggunakan plot pencar data berpasangan nilai autokorelasi dapat kita lihat menggunaka **plot ACF (*Autocorelation Function*) contoh**. Plot ACF contoh dapat kita gunakan untuk menentukan ordo dari proses MA.

```python
plot_acf(ma1, zero=False)
plt.show()
```

Bagaimana interpretasi plot ACF di atas?

<h1> Regresi Diri (Autoregressive) </h1>

Secara spesifik, proses autoregressive order p dari deret ${Y_t}$ memenuhi persamaan berikut:

$$
Y_t = \phi_1 Y_{t-1} + \phi_2 Y_{t-2} + \cdots + \phi_p Y_{t-p} + e_t
$$
Nilai $Y_t$ pada waktu sekarang merupakan kombinasi linier dari sebanyak p nilai $Y_t$ pada waktu lampau ditambah $e_t$ yang saling bebas terhadap $Y_{t-1}, Y_{t-2},...$.


<h2> AR (1) </h2>

Proses AR (1) memiliki persamaan:

$$
Y_t = \phi Y_{t-1} + e_t
$$
**Syarat kestasioneran untuk deret AR (1) adalah jika dan hanya jika $|\phi|<1$.**

Dari plot ACF contoh tampak nilai autokorelasi yang cukup tinggi dan signifikan sampai lag 6. Perhatikan bahwa nilai autokorelasi bergantian tanda dari negatif ke positif.

**Karena $|\phi|<1$, nilai fungsi autokorelasi berkurang secara eksponensial seiring dengan meningkatnya jumlah lag k. Jika $0<\phi<1$, semua korelasinya bernilai positif, dan jika $-1<\phi<0$, autokorelasi pada lag 1 bernilai negatif dan tanda untuk autokorelasi yang berurutan bergantian dari positif ke negatif, dengan nilainya berkurang secara eksponensial.**

Berdasarkan sifat tersebut, maka nilai ACF contoh tidak dapat kita gunakan untuk menentukan ordo dari proses AR, sehingga kita butuh alat lain. **PACF (Partial Autocorelation Function) contoh** merupakan salah satu alat yang bisa kita gunakan untuk menentukan ordo dari model AR.

<h2> AR (2) </h2>

Proses AR (2) memiliki persamaan:

$$
Y_t = \phi_1 Y_{t-1} + \phi_2 Y_{t-2} + e_t
$$
**Syarat kestasioneran untuk deret AR (2) adalah jika dan hanya jika tiga kondisi berikut terpenuhi, yaitu:**

$$
\phi_1 + \phi_2 < 1, \qquad
\phi_2 - \phi_1 < 1, \qquad
|\phi_2| < 1
$$
Kita simulasikan deret AR (2) dengan $\phi_1=1.5$ dan $\phi_2=-0.75$

```python
plt.figure(figsize=(10,5))
plt.plot(ar2,'-o')
plt.xlabel('Waktu'), plt.ylabel(r'$Y_t$')
plt.show()
```

```python
plot_acf(np.array(ar2), zero=False)
plt.show()
```

Perhatikan pola ACF tersebut. Apa kaitan pola yang terbentuk dengan sifat dari proses AR?

```python
plot_pacf(np.array(ar2), zero=False)
plt.show()
```

<h1> Model Campuran ARMA </h1>

Jika diasumsikan bahwa deret terdiri dari setengah autoregressive dan setengah lagi rataan bergerak, akan diperoleh model deret waktu yang cukup umum dengan persamaan:

$$
\begin{aligned}
Y_t ={}& \phi_1 Y_{t-1} + \phi_2 Y_{t-2} + \cdots + \phi_p Y_{t-p} \\
       &+ e_t - \theta_1 e_{t-1} - \theta_2 e_{t-2} - \cdots - \theta_q e_{t-q}
\end{aligned}
$$

Dikatakan bahwa $${Y_t} $$ merupakan proses campuran autoregressive moving average ordo p dan q, atau disingkat dengan ARMA(p,q).

<h2> ARMA (1,1) </h2>

Persamaan umum untuk proses ARMA (1,1) adalah:

$$
Y_t = \phi Y_{t-1} + e_t - \theta e_{t-1}
$$
Kita simulasikan deret ARMA (1,1) dengan $\phi=0.6$ dan $\theta=-0.3$

```python
np.random.seed(2026)
ar1 = np.array([1,-0.6]) #nilai parameter phi dalam sintaks dituliskan sebagai negatif dari nilai parameter asli
ma1 = np.array([1,0.3]) #nilai parameter theta dalam sintaks dituliskan sebagai negatif dari nilai parameter asli
arma_object1 = ArmaProcess(ar1,ma1)
arma1 = arma_object1.generate_sample(nsample=100)
```

```python
plt.figure(figsize=(10,5))
plt.plot(arma1,'-o')
plt.xlabel('Waktu'), plt.ylabel(r'$Y_t$')
plt.show()
```

Bagaimana pola dari deret tersebut?
Karakteristiknya:

Stasioner, karena (|\phi_1|=0{,}6<1).
Invertibel, karena (|\theta_1|=0{,}3<1).
Jika tidak ada konstanta dan (e_t) memiliki rataan nol, deret berfluktuasi di sekitar rataan nol.
ACF meluruh secara geometris: kira-kira (\rho_1=0{,}732), (\rho_2=0{,}439), dan (\rho_3=0{,}263). Jadi nilai berurutan cenderung berkorelasi positif, tetapi pengaruhnya makin kecil pada lag yang lebih jauh.
Baik ACF maupun PACF umumnya meluruh, bukan terputus tajam.

```python
plot_acf(arma1, zero=False)
plt.show()
```

```python
plot_pacf(arma1, zero=False)
plt.show()
```

Berdasarkan plot ACF contoh contoh terlihat bahwa model yang disarankan adalah MA (2), hal ini tentu tidak sesuai dengan spesifikasi awal deret, dimana untuk komponen rataan bergerak kita definisikan proses MA (1). Sementara dari plot PACF contoh model yang disarankan adalah AR (1). Sehingga model yang sesuai berdasarkan plot ACF dan PACF contoh adalah model AR (1) atau MA (2), dimana seharusnya model yang digunakan adalah ARMA (1,1). Hal ini bisa saja terjadi, sehingga dalam pemodelan deret waktu biasanya digunakan beberapa model tentatif dengan melakukan *overfitting* atau *underfitting* dari model yang disarankan oleh ACF dan PACF.

Kita simulasikan deret ARMA (1,1) yang lainnya secara manual berdasarkan persamaan proses ARMA (1,1) dengan $\phi=0.75$ dan $\theta=0.4$

```python
# Jawaban Soal 1
# Simulasi Data MA(1) dengan berbagai nilai parameter theta
import numpy as np
import matplotlib.pyplot as plt

n = 20
theta_values = [0.1, 0.3, 0.7, 1.0, 1.5, 2.0]

np.random.seed(120)
e = np.random.normal(loc=0, scale=1, size=n + 1)

fig, axes = plt.subplots(3, 2, figsize=(12, 9), sharex=True)
waktu = np.arange(1, n + 1)

for ax, theta in zip(axes.flat, theta_values):
    # Model MA(1): Y_t = e_t - theta * e_(t-1)
    y = e[1:] - theta * e[:-1]

    ax.plot(waktu, y, marker="o")
    ax.axhline(0, color="gray", linewidth=0.8)
    ax.set_title(f"MA(1), θ = {theta}")
    ax.set_xlabel("Waktu")
    ax.set_ylabel("$Y_t$")
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Jawaban Soal 2
import numpy as np
import matplotlib.pyplot as plt

n = 20
theta_values = [0.1, 0.3, 0.7, 1.0, 1.5, 2.0]
max_lag = 10

# Inovasi yang sama dipakai untuk semua nilai theta
np.random.seed(120)
e = np.random.normal(loc=0, scale=1, size=n + 1)

fig, axes = plt.subplots(3, 2, figsize=(12, 10), sharex=True, sharey=True)
lags = np.arange(max_lag + 1)
batas = 1.96 / np.sqrt(n)  # batas signifikansi pendekatan 95%

for ax, theta in zip(axes.flat, theta_values):
    # Model MA(1): Y_t = e_t - theta * e_(t-1)
    y = e[1:] - theta * e[:-1]

    # ACF sampel, dengan lag 0 sampai max_lag
    y_centered = y - y.mean()
    acf_full = np.correlate(y_centered, y_centered, mode="full")[n - 1:]
    acf = acf_full[:max_lag + 1] / acf_full[0]

    # ACF teoretis lag 1 untuk model Y_t = e_t - theta * e_(t-1)
    rho1_teori = -theta / (1 + theta**2)

    ax.stem(lags, acf, linefmt="C0-", markerfmt="C0o", basefmt="k-")
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axhline(batas, color="gray", linestyle="--", linewidth=0.8)
    ax.axhline(-batas, color="gray", linestyle="--", linewidth=0.8)
    ax.plot(1, rho1_teori, marker="D", color="red",
            label=fr"Teori lag 1: {rho1_teori:.3f}")

    ax.set_title(fr"MA(1), $\theta={theta}$")
    ax.set_xlabel("Lag")
    ax.set_ylabel("ACF")
    ax.set_xticks(lags)
    ax.grid(True, alpha=0.3)
    ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```

Secara teori, untuk model tersebut:

$$
\rho_1=-\frac{\theta}{1+\theta^2},
\qquad
\rho_k=0 \quad \text{untuk } k>1.
$$

Jadi, untuk semua nilai (\theta) positif, autokorelasi teoretis pada lag 1 negatif. Besar korelasinya meningkat hingga (\theta=1), saat (\rho_1=-0,5), lalu menurun lagi. Karena deret hanya memiliki 20 pengamatan, ACF sampelnya bisa berbeda cukup jauh dari nilai teori, dan lag di atas 1 tidak harus tepat nol.

```python
# Jawaban Soal 4    
import numpy as np
import matplotlib.pyplot as plt

theta = 0.25
panjang_pengamatan = [60, 100, 150, 300]
max_lag = 20

# Inovasi yang sama, sehingga deret lebih pendek menjadi awalan deret lebih panjang
np.random.seed(120)
e = np.random.normal(0, 1, size=max(panjang_pengamatan) + 1)

# ACF teoretis MA(1) dengan model Y_t = e_t - theta * e_(t-1)
rho1_teori = -theta / (1 + theta**2)

fig, axes = plt.subplots(2, 2, figsize=(12, 9), sharex=True, sharey=True)
lags = np.arange(max_lag + 1)

for ax, n in zip(axes.flat, panjang_pengamatan):
    y = e[1:n + 1] - theta * e[:n]
    y_centered = y - y.mean()

    # ACF sampel
    denominator = np.dot(y_centered, y_centered)
    acf = np.array([
        np.dot(y_centered[:n - lag], y_centered[lag:]) / denominator
        if lag > 0 else 1.0
        for lag in lags
    ])

    # Batas pendekatan interval kepercayaan 95% di bawah hipotesis ACF = 0
    batas = 1.96 / np.sqrt(n)

    ax.stem(lags, acf, linefmt="C0-", markerfmt="C0o", basefmt="k-")
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axhspan(-batas, batas, color="gray", alpha=0.2,
               label=fr"95%: $\pm{batas:.3f}$")
    ax.plot(1, rho1_teori, "rD",
            label=fr"Teori lag 1: {rho1_teori:.3f}")

    ax.set_title(f"Panjang pengamatan = {n}")
    ax.set_xlabel("Lag")
    ax.set_ylabel("ACF")
    ax.set_xticks(lags)
    ax.grid(True, alpha=0.3)
    ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```

Secara teori, autokorelasi lag 1 model ini adalah

$$
\rho_1 = -\frac{\theta}{1+\theta^2}
= -\frac{0.25}{1+0.25^2}
\approx -0.235
$$

sedangkan autokorelasi pada lag di atas 1 adalah nol. Batas kepercayaan pendekatan 95% menyempit seiring bertambahnya jumlah pengamatan: sekitar (\pm0.253), (\pm0.196), (\pm0.160), dan (\pm0.113) untuk panjang 60, 100, 150, dan 300. Jadi, estimasi ACF lag 1 biasanya mendekati nilai teori negatif; dengan sampel lebih besar, estimasi itu lebih jelas berada di luar batas nol. Sampel kecil dapat menghasilkan ACF yang cukup berbeda dari teori.

```python
# Jawaban Soal 5    
import numpy as np
import matplotlib.pyplot as plt

n = 50
parameter = [
    (0.3, 0.8),
    (-0.3, 0.8),
    (0.3, -0.8),
    (-0.3, -0.8),
]

np.random.seed(120)
e = np.random.normal(loc=0, scale=1, size=n + 2)

fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True)

for ax, (theta1, theta2) in zip(axes.flat, parameter):
    # Model MA(2): Y_t = e_t - theta1*e_(t-1) - theta2*e_(t-2)
    y = e[2:] - theta1 * e[1:-1] - theta2 * e[:-2]

    ax.plot(np.arange(1, n + 1), y, marker="o", markersize=3)
    ax.axhline(0, color="gray", linewidth=0.8)
    ax.set_title(
        rf"$\theta_1={theta1},\ \theta_2={theta2}$"
    )
    ax.set_xlabel("Waktu")
    ax.set_ylabel("$Y_t$")
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

Untuk model yang digunakan, (Y_t=e_t-\theta_1e_{t-1}-\theta_2e_{t-2}), tanda parameter memengaruhi autokorelasi. Secara teori, autokorelasi setelah lag 2 bernilai nol, sedangkan nilainya pada lag 1 dan 2 adalah:

((\theta_1,\theta_2))	(\rho_1)	(\rho_2)
((0.3, 0.8))	(-0.035)	(-0.462)
((-0.3, 0.8))	(0.035)	(-0.462)
((0.3, -0.8))	(-0.312)	(0.462)
((-0.3, -0.8))	(0.312)	(0.462)
Kesimpulan: tanda (\theta_1) membalik tanda autokorelasi lag 1, sementara tanda (\theta_2) membalik tanda autokorelasi lag 2. Untuk melihat pola tersebut dengan lebih jelas, bandingkan plot ACF masing-masing deret. Sampel sepanjang 50 masih bisa menunjukkan penyimpangan acak dari nilai teoretis.

```python
# Jawaban Soal 6
import numpy as np
import matplotlib.pyplot as plt

n = 50
max_lag = 15
parameter = [
    (0.3, 0.8),
    (-0.3, 0.8),
    (0.3, -0.8),
    (-0.3, -0.8),
]

np.random.seed(120)
e = np.random.normal(0, 1, size=n + 2)

fig, axes = plt.subplots(2, 2, figsize=(12, 9), sharex=True, sharey=True)
lags = np.arange(max_lag + 1)
batas = 1.96 / np.sqrt(n)

for ax, (theta1, theta2) in zip(axes.flat, parameter):
    # MA(2): Y_t = e_t - theta1*e_(t-1) - theta2*e_(t-2)
    y = e[2:] - theta1 * e[1:-1] - theta2 * e[:-2]
    y = y - y.mean()

    penyebut = np.dot(y, y)
    acf = np.array([
        np.dot(y[:n - lag], y[lag:]) / penyebut
        if lag > 0 else 1.0
        for lag in lags
    ])

    # ACF teoretis untuk model dengan tanda minus
    penyebut_teori = 1 + theta1**2 + theta2**2
    rho1 = theta1 * (theta2 - 1) / penyebut_teori
    rho2 = -theta2 / penyebut_teori

    ax.stem(lags, acf, linefmt="C0-", markerfmt="C0o", basefmt="k-")
    ax.axhline(0, color="black", linewidth=0.8)
    ax.axhline(batas, color="gray", linestyle="--", linewidth=0.8)
    ax.axhline(-batas, color="gray", linestyle="--", linewidth=0.8)
    ax.plot([1, 2], [rho1, rho2], "rD", label="ACF teori (lag 1–2)")

    ax.set_title(rf"$\theta_1={theta1},\ \theta_2={theta2}$")
    ax.set_xlabel("Lag")
    ax.set_ylabel("ACF")
    ax.set_xticks(lags)
    ax.grid(True, alpha=0.3)
    ax.legend(fontsize=8)

plt.tight_layout()
plt.show()
```

Kesimpulan: ACF teoretis MA(2) dapat bernilai bukan nol pada lag 1 dan 2, lalu terputus setelah lag 2. Tanda parameter memengaruhi arah korelasi, sedangkan besar nilainya juga bergantung pada kedua parameter. Pada plot sampel sepanjang 50, lag di atas 2 mungkin tampak tidak nol karena variasi acak; garis putus-putus menunjukkan batas kepercayaan pendekatan 95% untuk autokorelasi nol.

```python
# Jawaban Soal 7
import numpy as np
import matplotlib.pyplot as plt

n = 100
phi_values = [0.3, 0.9, -0.3, -0.9]

np.random.seed(120)
e = np.random.normal(loc=0, scale=1, size=n)

fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True)

for ax, phi in zip(axes.flat, phi_values):
    y = np.zeros(n)

    for t in range(1, n):
        y[t] = phi * y[t - 1] + e[t]

    ax.plot(np.arange(1, n + 1), y)
    ax.axhline(0, color="gray", linewidth=0.8)
    ax.set_title(rf"AR(1), $\phi={phi}$")
    ax.set_xlabel("Waktu")
    ax.set_ylabel("$Y_t$")
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

Semua nilai (\phi) tersebut memenuhi (|\phi|<1), sehingga proses AR(1) secara teori stasioner. Nilai (\phi) positif cenderung menghasilkan pergerakan berurutan yang searah; nilai negatif cenderung menghasilkan pola naik-turun bergantian. Semakin dekat (|\phi|) ke 1, semakin kuat dan lambat meredanya ketergantungan antarwaktu.

```python
# Jawaban Soal 8
import numpy as np
import matplotlib.pyplot as plt

n = 100
max_lag = 20
phi_values = [0.3, 0.9, -0.3, -0.9]

# Inovasi yang sama untuk setiap nilai phi
np.random.seed(120)
e = np.random.normal(loc=0, scale=1, size=n)

def hitung_acf(y, max_lag):
    """Menghitung ACF sampel dari lag 0 hingga max_lag."""
    y = y - y.mean()
    penyebut = np.dot(y, y)

    return np.array([
        np.dot(y[:len(y) - lag], y[lag:]) / penyebut
        if lag > 0 else 1.0
        for lag in range(max_lag + 1)
    ])

def hitung_pacf_dari_acf(acf, max_lag):
    """Menghitung PACF melalui persamaan Yule-Walker."""
    pacf = np.empty(max_lag + 1)
    pacf[0] = 1.0

    for k in range(1, max_lag + 1):
        matriks = np.array([
            [acf[abs(i - j)] for j in range(k)]
            for i in range(k)
        ])
        ruas_kanan = acf[1:k + 1]
        koefisien = np.linalg.solve(matriks, ruas_kanan)
        pacf[k] = koefisien[-1]

    return pacf

fig, axes = plt.subplots(4, 2, figsize=(12, 14), sharex=True)
lags = np.arange(max_lag + 1)
batas = 1.96 / np.sqrt(n)

for row, phi in enumerate(phi_values):
    # AR(1): Y_t = phi * Y_(t-1) + e_t
    y = np.zeros(n)
    for t in range(1, n):
        y[t] = phi * y[t - 1] + e[t]

    acf = hitung_acf(y, max_lag)
    pacf = hitung_pacf_dari_acf(acf, max_lag)

    for col, values, nama in [
        (0, acf, "ACF"),
        (1, pacf, "PACF"),
    ]:
        ax = axes[row, col]
        ax.stem(lags, values, linefmt="C0-", markerfmt="C0o", basefmt="k-")
        ax.axhline(0, color="black", linewidth=0.8)
        ax.axhline(batas, color="gray", linestyle="--", linewidth=0.8)
        ax.axhline(-batas, color="gray", linestyle="--", linewidth=0.8)
        ax.set_title(f"{nama}, AR(1) dengan φ = {phi}")
        ax.set_ylabel(nama)
        ax.set_xticks(lags)
        ax.grid(True, alpha=0.3)

        if row == len(phi_values) - 1:
            ax.set_xlabel("Lag")

plt.tight_layout()
plt.show()
```

Kesimpulan: Untuk AR(1), ACF secara teori meluruh bertahap mengikuti (\rho_k=\phi^k). Jika (\phi) positif, ACF tetap positif dan meluruh; jika (\phi) negatif, tandanya berganti-ganti. Semakin dekat (|\phi|) ke 1, semakin lambat peluruhannya. PACF secara teori menunjukkan korelasi utama pada lag 1 dan mendekati nol setelahnya. Pada sampel terbatas, PACF pada lag lain bisa tampak sedikit berbeda dari nol karena variasi acak.

Berdasarkan pola ACF dan PACF, model yang disarankan untuk **keempat deret adalah AR(1)**, karena ACF meluruh bertahap, sementara PACF terutama menunjukkan lonjakan pada lag 1 lalu mendekati nol.

| Deret | Model yang disarankan | Pola yang diharapkan |

| \(\phi=0.3\) | AR(1), \(Y_t=0.3Y_{t-1}+e_t\) | ACF positif dan meluruh relatif cepat |
| \(\phi=0.9\) | AR(1), \(Y_t=0.9Y_{t-1}+e_t\) | ACF positif dan meluruh lambat |
| \(\phi=-0.3\) | AR(1), \(Y_t=-0.3Y_{t-1}+e_t\) | ACF berganti tanda dan meluruh relatif cepat |
| \(\phi=-0.9\) | AR(1), \(Y_t=-0.9Y_{t-1}+e_t\) | ACF berganti tanda dan meluruh lambat |

Tanda \(\phi\) menentukan apakah korelasi antar-lag cenderung searah atau berganti arah; besar \(|\phi|\) menentukan seberapa lambat korelasinya meluruh. PACF sampel mungkin tidak tepat nol setelah lag 1 karena deret hanya memiliki 100 pengamatan.

```python
# Jawaban Soal 10
import numpy as np
import matplotlib.pyplot as plt

n = 100
parameter = [
    (1.5, -0.5),
    (0.75, -0.25),
    (1.0, -0.25),
    (0.5, -0.75),
]

np.random.seed(120)
e = np.random.normal(loc=0, scale=1, size=n)

fig, axes = plt.subplots(2, 2, figsize=(12, 8), sharex=True)

for ax, (phi1, phi2) in zip(axes.flat, parameter):
    y = np.zeros(n)

    for t in range(n):
        y_lag1 = y[t - 1] if t >= 1 else 0
        y_lag2 = y[t - 2] if t >= 2 else 0
        y[t] = phi1 * y_lag1 + phi2 * y_lag2 + e[t]

    ax.plot(np.arange(1, n + 1), y)
    ax.axhline(0, color="gray", linewidth=0.8)
    ax.set_title(rf"AR(2), $\phi_1={phi1},\ \phi_2={phi2}$")
    ax.set_xlabel("Waktu")
    ax.set_ylabel("$Y_t$")
    ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

Catatan: untuk ((\phi_1,\phi_2)=(1.5,-0.5)), (\phi_1+\phi_2=1), sehingga model berada pada batas stasioner dan memiliki akar unit. Tiga kombinasi parameter lainnya memenuhi syarat stasioneritas AR(2).

```python
# Jawaban Soal 11
import numpy as np
import matplotlib.pyplot as plt

n = 100
max_lag = 20
parameter = [
    (1.5, -0.5),
    (0.75, -0.25),
    (1.0, -0.25),
    (0.5, -0.75),
]

# Gunakan inovasi yang sama untuk semua kombinasi parameter
np.random.seed(120)
e = np.random.normal(loc=0, scale=1, size=n)

def hitung_acf(y, max_lag):
    y = y - y.mean()
    penyebut = np.dot(y, y)

    return np.array([
        np.dot(y[:len(y) - lag], y[lag:]) / penyebut
        if lag > 0 else 1.0
        for lag in range(max_lag + 1)
    ])

def hitung_pacf_yule_walker(acf, max_lag):
    pacf = np.empty(max_lag + 1)
    pacf[0] = 1.0

    for k in range(1, max_lag + 1):
        matriks = np.array([
            [acf[abs(i - j)] for j in range(k)]
            for i in range(k)
        ])
        koefisien = np.linalg.solve(matriks, acf[1:k + 1])
        pacf[k] = koefisien[-1]

    return pacf

fig, axes = plt.subplots(4, 2, figsize=(12, 14), sharex=True)
lags = np.arange(max_lag + 1)
batas = 1.96 / np.sqrt(n)

for baris, (phi1, phi2) in enumerate(parameter):
    # Simulasi AR(2): Y_t = phi1*Y_(t-1) + phi2*Y_(t-2) + e_t
    y = np.zeros(n)

    for t in range(n):
        y_lag1 = y[t - 1] if t >= 1 else 0
        y_lag2 = y[t - 2] if t >= 2 else 0
        y[t] = phi1 * y_lag1 + phi2 * y_lag2 + e[t]

    acf = hitung_acf(y, max_lag)
    pacf = hitung_pacf_yule_walker(acf, max_lag)

    for kolom, nilai, nama in [
        (0, acf, "ACF"),
        (1, pacf, "PACF"),
    ]:
        ax = axes[baris, kolom]
        ax.stem(lags, nilai, linefmt="C0-", markerfmt="C0o", basefmt="k-")
        ax.axhline(0, color="black", linewidth=0.8)
        ax.axhline(batas, color="gray", linestyle="--", linewidth=0.8)
        ax.axhline(-batas, color="gray", linestyle="--", linewidth=0.8)
        ax.set_title(
            rf"{nama}, AR(2): $\phi_1={phi1},\ \phi_2={phi2}$"
        )
        ax.set_ylabel(nama)
        ax.set_xticks(lags)
        ax.grid(True, alpha=0.3)

        if baris == len(parameter) - 1:
            ax.set_xlabel("Lag")

plt.tight_layout()
plt.show()
```

Kesimpulan umum: pada proses AR(2) yang stasioner, ACF biasanya meluruh bertahap, sedangkan PACF menunjukkan lonjakan utama hingga lag 2 lalu mendekati nol. Tanda dan besar (\phi_1,\phi_2) memengaruhi apakah ACF meluruh secara monoton atau berosilasi, serta seberapa cepat peluruhannya.
